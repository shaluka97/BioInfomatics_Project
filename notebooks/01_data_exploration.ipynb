# 01 — Data Exploration
Explore the simulator output: case/control balance, MAF distribution, within-block LD, and where the planted causal SNPs sit.

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))
import numpy as np, matplotlib.pyplot as plt
from src.data_simulation import SimulationConfig, simulate_dataset
cfg = SimulationConfig(n_individuals=1000, n_snps=5000, seed=42)
X, y, truth = simulate_dataset(cfg)
print('X:', X.shape, 'case rate:', y.mean())
print('strong/weak/epistatic-flat:', truth['strong'].size, truth['weak'].size, truth['epistatic_flat'].size)
```

```python
maf = np.minimum(X.mean(0)/2, 1 - X.mean(0)/2)
plt.hist(maf, bins=40); plt.xlabel('MAF'); plt.ylabel('# SNPs'); plt.title('MAF distribution'); plt.show()
```

```python
Xz = (X - X.mean(0)) / (X.std(0) + 1e-6)
block = Xz[:, :50]; corr = (block.T @ block) / X.shape[0]
import seaborn as sns; sns.heatmap(corr, vmin=-1, vmax=1, cmap='vlag'); plt.title('Within-block LD (block 0)'); plt.show()
```
